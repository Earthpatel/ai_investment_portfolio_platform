```python
# -----------------------------------
# Module 4: Modeling & Prediction
# Linear Regression for Stock Returns
# -----------------------------------

import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# -----------------------------------
# 1️⃣ Load Data
# -----------------------------------

df = pd.read_csv("../data/processed/features_dataset.csv.csv")

# Ensure proper sorting (VERY IMPORTANT)
df = df.sort_values(by=["Ticker", "Date"])

# Convert Date to datetime
df["Date"] = pd.to_datetime(df["Date"])

# Drop missing values
df = df.dropna()

# -----------------------------------
# 2️⃣ Define Target (y)
# -----------------------------------

# Example: Predict next-day return
# (Make sure this column exists from your feature engineering module)
target_col = "Next_Return"

# -----------------------------------
# 3️⃣ Define Features (X)
# -----------------------------------

feature_cols = [
    "Close",
]

# -----------------------------------
# 4️⃣ OPTIONAL: Encode Ticker (Recommended)
# -----------------------------------

df = pd.get_dummies(df, columns=["Ticker"], drop_first=True)

# Update feature columns to include ticker dummies
feature_cols = [col for col in df.columns if col not in ["Date", target_col]]

# -----------------------------------
# 5️⃣ Train/Test Split (Time-Based)
# -----------------------------------

# Split by date (prevents data leakage)
split_date = df["Date"].quantile(0.8)

train_df = df[df["Date"] <= split_date]
test_df  = df[df["Date"] > split_date]

X_train = train_df[feature_cols]
y_train = train_df[target_col]

X_test  = test_df[feature_cols]
y_test  = test_df[target_col]

# -----------------------------------
# 6️⃣ Train Model
# -----------------------------------

model = LinearRegression()
model.fit(X_train, y_train)

# -----------------------------------
# 7️⃣ Predictions
# -----------------------------------

y_pred = model.predict(X_test)

# -----------------------------------
# 8️⃣ Evaluation
# -----------------------------------

rmse = mean_squared_error(y_test, y_pred, squared=False)
mae  = mean_absolute_error(y_test, y_pred)
r2   = r2_score(y_test, y_pred)

print("\n📊 Model Performance:")
print("RMSE:", rmse)
print("MAE :", mae)
print("R²  :", r2)

# -----------------------------------
# 9️⃣ Coefficient Interpretation
# -----------------------------------

coeff_df = pd.DataFrame({
    "Feature": X_train.columns,
    "Coefficient": model.coef_
}).sort_values(by="Coefficient", ascending=False)

print("\n📈 Feature Importance (Linear Coefficients):")
print(coeff_df.head(15))

print("\nIntercept:", model.intercept_)

# -----------------------------------
# 🔟 Save Predictions
# -----------------------------------

test_df = test_df.copy()
test_df["Predicted_Return"] = y_pred

# Direction signal (important for trading logic)
test_df["Predicted_Direction"] = (test_df["Predicted_Return"] > 0).astype(int)

# Save output
test_df.to_csv("../data/output/predictions_linear.csv", index=False)

# -----------------------------------
# 1️⃣1️⃣ (Optional) Simple Backtest Signal
# -----------------------------------

# Strategy: Invest only when model predicts positive return
test_df["Strategy_Return"] = test_df["Predicted_Direction"] * test_df[target_col]

# Compare cumulative returns
test_df["Cumulative_Market"] = (1 + test_df[target_col]).cumprod()
test_df["Cumulative_Strategy"] = (1 + test_df["Strategy_Return"]).cumprod()

print("\n📊 Final Strategy vs Market:")
print(test_df[["Cumulative_Market", "Cumulative_Strategy"]].tail())

# -----------------------------------
# END
# -----------------------------------
```
