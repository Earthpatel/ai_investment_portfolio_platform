# 01 — Data Collection
**AI Investment Manager Platform**

This notebook fetches S&P 500 price history, company info, and financials from yfinance  
and saves clean, validated Parquet files ready for feature engineering.

### Key design decisions
- **Pinned date range** — reproducible across re-runs
- **All 503 tickers** — batched with rate-limit pauses
- **Three separate tables** — history / info / financials (no premature merge)
- **Parquet output** — 10× faster than CSV at this scale
- **Quality checks** — null rates, shape assertions, dupe detection before saving

## 0. Imports & Configuration

```python
import yfinance as yf
import pandas as pd
import numpy as np
import os
import time
import logging
from pathlib import Path
from datetime import datetime

# ── Directory setup (MUST come before logging so the logs/ folder exists) ─────
DIRS = [
    '../data/raw',
    '../data/processed',
    '../logs'
]
for d in DIRS:
    Path(d).mkdir(parents=True, exist_ok=True)

# ── Logging setup ──────────────────────────────────────────────────────────────
logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s | %(levelname)s | %(message)s',
    handlers=[
        logging.FileHandler('../logs/data_collection.log'),
        logging.StreamHandler()
    ]
)
logger = logging.getLogger(__name__)

# ── Pinned date range (change end date when retraining) ───────────────────────
# Pinning prevents training data from silently shifting between runs.
START_DATE = '2020-01-01'
END_DATE   = '2025-01-01'

# ── Fetch config ───────────────────────────────────────────────────────────────
BATCH_SIZE       = 50    # tickers per batch
SLEEP_BETWEEN    = 2.0   # seconds between batches (avoid rate limits)
MAX_NULL_RATE    = 0.10  # reject any column with >10% nulls in history
MIN_TRADING_DAYS = 200   # drop tickers with fewer rows than this

logger.info(f'Config loaded. Date range: {START_DATE} → {END_DATE}')
```

## 1. Load Tickers

```python
FILE_PATH = '../data/raw/sp500_tickers.csv'

if not os.path.exists(FILE_PATH):
    raise FileNotFoundError(
        f'Cannot find {FILE_PATH}. '
        'Run: pd.read_html("https://en.wikipedia.org/wiki/List_of_S%26P_500_companies")[0]'
        '.to_csv(FILE_PATH, index=False)  to regenerate it.'
    )

tickers_df = pd.read_csv(FILE_PATH)

# Support both 'Symbol' and 'Ticker' column names
col = 'Symbol' if 'Symbol' in tickers_df.columns else 'Ticker'

# Normalise BRK.B → BRK-B (yfinance uses hyphens)
tickers = tickers_df[col].str.strip().str.replace('.', '-', regex=False).tolist()

logger.info(f'Loaded {len(tickers)} tickers.')
print(f'Total tickers: {len(tickers)}')
print(f'Sample: {tickers[:10]}')
```

## 2. Fetch Price History — All 503 Tickers

We iterate in batches of 50 with a 2-second pause between batches.  
Failed tickers are logged and skipped — they won't silently corrupt the dataset.

```python
def fetch_history_batch(ticker_list: list, start: str, end: str) -> tuple[pd.DataFrame, list]:
    """
    Fetch OHLCV history for a list of tickers.
    Returns (dataframe, failed_tickers).
    """
    frames   = []
    failed   = []

    for ticker in ticker_list:
        try:
            df = yf.Ticker(ticker).history(start=start, end=end)

            if df.empty:
                logger.warning(f'No history data returned for {ticker}')
                failed.append(ticker)
                continue

            if len(df) < MIN_TRADING_DAYS:
                logger.warning(f'{ticker}: only {len(df)} rows — skipping (< {MIN_TRADING_DAYS})')
                failed.append(ticker)
                continue

            df = df.reset_index()
            df['Ticker'] = ticker

            # Keep only essential columns
            df = df[['Ticker', 'Date', 'Open', 'High', 'Low', 'Close', 'Volume']]

            # Normalise timezone → plain date for clean joins later
            df['Date'] = pd.to_datetime(df['Date']).dt.tz_localize(None).dt.normalize()

            frames.append(df)

        except Exception as e:
            logger.error(f'Error fetching history for {ticker}: {e}')
            failed.append(ticker)

    combined = pd.concat(frames, ignore_index=True) if frames else pd.DataFrame()
    return combined, failed


# ── Run in batches ─────────────────────────────────────────────────────────────
all_history_frames = []
all_failed_history = []

total_batches = (len(tickers) + BATCH_SIZE - 1) // BATCH_SIZE

for batch_num, i in enumerate(range(0, len(tickers), BATCH_SIZE), start=1):
    batch = tickers[i : i + BATCH_SIZE]
    logger.info(f'History batch {batch_num}/{total_batches} — tickers {i}–{i+len(batch)-1}')

    batch_df, batch_failed = fetch_history_batch(batch, START_DATE, END_DATE)

    if not batch_df.empty:
        all_history_frames.append(batch_df)
    all_failed_history.extend(batch_failed)

    if batch_num < total_batches:
        time.sleep(SLEEP_BETWEEN)

all_history = pd.concat(all_history_frames, ignore_index=True)

logger.info(f'History fetch complete. Shape: {all_history.shape}')
logger.info(f'Failed tickers: {all_failed_history}')
print(f'\nHistory shape: {all_history.shape}')
print(f'Failed / skipped: {len(all_failed_history)} tickers')
print(all_history.head())
```

## 3. Validate Price History

```python
def validate_history(df: pd.DataFrame) -> None:
    """Run assertions and print a quality report on the history table."""
    print('=' * 55)
    print('HISTORY QUALITY REPORT')
    print('=' * 55)

    assert not df.empty,                          'FAIL: History DataFrame is empty'
    assert 'Close' in df.columns,                 'FAIL: Missing Close column'
    assert 'Ticker' in df.columns,                'FAIL: Missing Ticker column'
    assert df['Close'].isnull().mean() < 0.05,    'FAIL: >5% nulls in Close'
    assert df.duplicated(['Ticker','Date']).sum() == 0, 'FAIL: Duplicate Ticker+Date rows found'

    # Null rates
    null_rates = df.isnull().mean().sort_values(ascending=False)
    print('\nNull rates per column:')
    print(null_rates[null_rates > 0].to_string() if null_rates.any() else '  None — clean!')

    # Date coverage
    print(f'\nDate range : {df["Date"].min().date()} → {df["Date"].max().date()}')
    print(f'Total rows : {len(df):,}')
    print(f'Tickers    : {df["Ticker"].nunique()}')

    # Rows per ticker stats
    rows_per_ticker = df.groupby('Ticker').size()
    print(f'\nRows per ticker — min: {rows_per_ticker.min()}  '
          f'max: {rows_per_ticker.max()}  '
          f'median: {rows_per_ticker.median():.0f}')

    # Tickers with unusually few rows
    thin = rows_per_ticker[rows_per_ticker < MIN_TRADING_DAYS]
    if not thin.empty:
        print(f'\n⚠  Tickers with < {MIN_TRADING_DAYS} rows:')
        print(thin.to_string())
    else:
        print(f'\n✓  All tickers have ≥ {MIN_TRADING_DAYS} rows')

    print('\n✓  All assertions passed')
    print('=' * 55)

validate_history(all_history)
```

## 4. Fetch Company Info — All Tickers

```python
# Columns we actually need for ML feature engineering downstream.
# Keeping all 182 columns wastes memory and makes later joins messy.
INFO_COLS = [
    'Ticker',
    'shortName',
    'sector',
    'industry',
    'country',
    'marketCap',
    'trailingPE',
    'forwardPE',
    'priceToBook',
    'trailingEps',
    'forwardEps',
    'dividendYield',
    'beta',
    'fiftyTwoWeekHigh',
    'fiftyTwoWeekLow',
    'averageVolume',
    'totalDebt',
    'totalCash',
    'returnOnEquity',
    'returnOnAssets',
    'revenueGrowth',
    'grossMargins',
    'operatingMargins',
    'profitMargins',
    'debtToEquity',
    'currentRatio',
    'quickRatio',
    'sharesOutstanding',
    'floatShares',
    'heldPercentInsiders',
    'heldPercentInstitutions',
]


def fetch_info_batch(ticker_list: list) -> tuple[pd.DataFrame, list]:
    """
    Fetch company metadata for a list of tickers.
    Returns (dataframe, failed_tickers).
    """
    rows   = []
    failed = []

    for ticker in ticker_list:
        try:
            info = yf.Ticker(ticker).info
            if not info:
                logger.warning(f'Empty info for {ticker}')
                failed.append(ticker)
                continue

            row = {'Ticker': ticker}
            for col in INFO_COLS:
                if col != 'Ticker':
                    row[col] = info.get(col, np.nan)
            rows.append(row)

        except Exception as e:
            logger.error(f'Error fetching info for {ticker}: {e}')
            failed.append(ticker)

    df = pd.DataFrame(rows) if rows else pd.DataFrame()
    return df, failed


# ── Run in batches ─────────────────────────────────────────────────────────────
all_info_frames = []
all_failed_info  = []

for batch_num, i in enumerate(range(0, len(tickers), BATCH_SIZE), start=1):
    batch = tickers[i : i + BATCH_SIZE]
    logger.info(f'Info batch {batch_num}/{total_batches}')

    batch_df, batch_failed = fetch_info_batch(batch)
    if not batch_df.empty:
        all_info_frames.append(batch_df)
    all_failed_info.extend(batch_failed)

    if batch_num < total_batches:
        time.sleep(SLEEP_BETWEEN)

all_info = pd.concat(all_info_frames, ignore_index=True)

logger.info(f'Info fetch complete. Shape: {all_info.shape}')
print(f'Info shape : {all_info.shape}')
print(f'Failed     : {len(all_failed_info)}')
print(all_info.head())
```

## 5. Validate Company Info

```python
def validate_info(df: pd.DataFrame) -> None:
    print('=' * 55)
    print('INFO QUALITY REPORT')
    print('=' * 55)

    assert not df.empty,             'FAIL: Info DataFrame is empty'
    assert 'Ticker' in df.columns,   'FAIL: Missing Ticker column'
    assert 'sector' in df.columns,   'FAIL: Missing sector column'
    assert df['Ticker'].is_unique,   'FAIL: Duplicate tickers in info'

    null_rates = df.isnull().mean().sort_values(ascending=False)
    print('\nNull rates per column (top 10):')
    print(null_rates.head(10).to_string())

    print(f'\nSector distribution:')
    print(df['sector'].value_counts().to_string())

    print(f'\nTotal companies : {len(df)}')
    print('\n✓  All assertions passed')
    print('=' * 55)

validate_info(all_info)
```

## 6. Fetch Financials — All Tickers

We keep financials **separate** from history.  
Merging now would create a cartesian product (each daily row × each fiscal year = data explosion).  
The correct join happens in `02_feature_engineering.ipynb` by aligning fiscal year-end dates to price dates.

```python
def fetch_financials_batch(ticker_list: list) -> tuple[pd.DataFrame, list]:
    """
    Fetch annual income statement for a list of tickers.
    Returns (dataframe, failed_tickers).
    Columns = financial metrics; rows = (Ticker, FiscalYear).
    """
    frames = []
    failed = []

    for ticker in ticker_list:
        try:
            raw = yf.Ticker(ticker).financials  # shape: metrics × fiscal_years

            if raw is None or raw.empty:
                logger.warning(f'No financials for {ticker}')
                failed.append(ticker)
                continue

            # Transpose: each row = one fiscal year
            df = raw.T.copy()
            df.index.name = 'FiscalYearEnd'
            df = df.reset_index()

            # Normalise fiscal year date (strip timezone if present)
            df['FiscalYearEnd'] = pd.to_datetime(df['FiscalYearEnd']).dt.tz_localize(None)

            df.insert(0, 'Ticker', ticker)
            frames.append(df)

        except Exception as e:
            logger.error(f'Error fetching financials for {ticker}: {e}')
            failed.append(ticker)

    combined = pd.concat(frames, ignore_index=True) if frames else pd.DataFrame()
    return combined, failed


# ── Run in batches ─────────────────────────────────────────────────────────────
all_fin_frames   = []
all_failed_fin   = []

for batch_num, i in enumerate(range(0, len(tickers), BATCH_SIZE), start=1):
    batch = tickers[i : i + BATCH_SIZE]
    logger.info(f'Financials batch {batch_num}/{total_batches}')

    batch_df, batch_failed = fetch_financials_batch(batch)
    if not batch_df.empty:
        all_fin_frames.append(batch_df)
    all_failed_fin.extend(batch_failed)

    if batch_num < total_batches:
        time.sleep(SLEEP_BETWEEN)

all_financials = pd.concat(all_fin_frames, ignore_index=True)

logger.info(f'Financials fetch complete. Shape: {all_financials.shape}')
print(f'Financials shape : {all_financials.shape}')
print(f'Failed           : {len(all_failed_fin)}')
print(all_financials.head())
```

## 7. Validate Financials

```python
def validate_financials(df: pd.DataFrame) -> None:
    print('=' * 55)
    print('FINANCIALS QUALITY REPORT')
    print('=' * 55)

    assert not df.empty,                  'FAIL: Financials DataFrame is empty'
    assert 'Ticker' in df.columns,        'FAIL: Missing Ticker column'
    assert 'FiscalYearEnd' in df.columns, 'FAIL: Missing FiscalYearEnd column'

    dupes = df.duplicated(['Ticker', 'FiscalYearEnd']).sum()
    assert dupes == 0, f'FAIL: {dupes} duplicate Ticker+FiscalYearEnd rows'

    null_rates = df.isnull().mean().sort_values(ascending=False)
    high_null = null_rates[null_rates > MAX_NULL_RATE]
    print(f'\nColumns with >{MAX_NULL_RATE*100:.0f}% nulls ({len(high_null)} columns):')
    if not high_null.empty:
        print(high_null.head(15).to_string())
    else:
        print('  None')

    rows_per_ticker = df.groupby('Ticker').size()
    print(f'\nFiscal years per ticker — min: {rows_per_ticker.min()}  '
          f'max: {rows_per_ticker.max()}  '
          f'median: {rows_per_ticker.median():.0f}')

    print(f'\nDate range : {df["FiscalYearEnd"].min().date()} → {df["FiscalYearEnd"].max().date()}')
    print(f'Total rows : {len(df):,}')
    print('\n✓  All assertions passed')
    print('=' * 55)

validate_financials(all_financials)
```

## 8. Save to Parquet

Why Parquet over CSV?
- **~10× faster** read/write at this scale
- **Preserves dtypes** — no silent datetime or float → string conversions
- **Column pruning** — downstream notebooks only load columns they need
- **Smaller files** — built-in compression

```python
SAVE_DIR = Path('../data/processed')


def save_parquet(df: pd.DataFrame, name: str) -> None:
    """
    Save a DataFrame to Parquet with a timestamp suffix for versioning.
    Also saves a latest/ symlink-style copy for easy downstream loading.
    """
    # Latest version (used by downstream notebooks)
    latest_path = SAVE_DIR / f'{name}.parquet'
    df.to_parquet(latest_path, index=False, compression='snappy')

    # Versioned snapshot (keeps history of runs)
    timestamp = datetime.now().strftime('%Y%m%d_%H%M')
    versioned_path = SAVE_DIR / f'{name}_{timestamp}.parquet'
    df.to_parquet(versioned_path, index=False, compression='snappy')

    size_mb = latest_path.stat().st_size / 1_048_576
    logger.info(f'Saved {latest_path.name}  ({size_mb:.2f} MB)')
    print(f'✓  {latest_path.name:35s} {len(df):>8,} rows × {len(df.columns):>3} cols  ({size_mb:.2f} MB)')


print('Saving datasets...\n')
save_parquet(all_history,    'all_history')
save_parquet(all_info,       'all_info')
save_parquet(all_financials, 'all_financials')

# Also save a failed-tickers log for inspection
failed_summary = pd.DataFrame({
    'Ticker'  : all_failed_history + all_failed_info + all_failed_fin,
    'Stage'   : (['history'] * len(all_failed_history) +
                 ['info']    * len(all_failed_info)    +
                 ['financials'] * len(all_failed_fin))
})
failed_summary.to_csv('../logs/failed_tickers.csv', index=False)
print(f'\nFailed tickers log: ../logs/failed_tickers.csv  ({len(failed_summary)} entries)')
```

## 9. Smoke Test — Reload & Verify

Always reload from disk and re-check. This catches silent write failures.

```python
print('Reloading from disk...\n')

h = pd.read_parquet(SAVE_DIR / 'all_history.parquet')
i = pd.read_parquet(SAVE_DIR / 'all_info.parquet')
f = pd.read_parquet(SAVE_DIR / 'all_financials.parquet')

# Quick shape + dtype check
for label, df in [('History', h), ('Info', i), ('Financials', f)]:
    print(f'{label:12s} | shape: {str(df.shape):>18s} | dtypes OK: {df.dtypes.astype(str).nunique() > 1}')

# Spot-check: AAPL should always be present
assert 'AAPL' in h['Ticker'].values,  'AAPL missing from history'
assert 'AAPL' in i['Ticker'].values,  'AAPL missing from info'
assert 'AAPL' in f['Ticker'].values,  'AAPL missing from financials'

print('\n✓  Smoke test passed — all three tables reload cleanly with AAPL present')
```
